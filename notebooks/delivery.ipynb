```python

import pandas as pd

df = pd.read_csv("../data/qatar_delivery_orders_clean.csv")

# Keep delivered orders for this validation
delivered = df[
    df["order_status"].eq("Delivered")
].copy()

# Check whether the late-delivery labels agree with delivery times
delivered["late_from_time"] = (
    pd.to_numeric(delivered["actual_delivery_min"], errors="coerce")
    > pd.to_numeric(delivered["promised_delivery_min"], errors="coerce")
)

print("Delivered orders:", len(delivered))

print("\nExisting late-delivery labels:")
print(delivered["late_delivery"].value_counts())

print("\nLate delivery based on actual vs promised time:")
print(delivered["late_from_time"].value_counts())

print("\nAgreement between existing and calculated labels:")
print(
    pd.crosstab(
        delivered["late_delivery"],
        delivered["late_from_time"],
        rownames=["Existing label"],
        colnames=["Calculated label"]
    )
)
```
